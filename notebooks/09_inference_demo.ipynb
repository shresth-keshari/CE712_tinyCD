# Local inference and evidence export
Inference-only walkthrough using frozen released checkpoints. See ../demo_guide.md and ../USGS_WORKFLOW.md.


```python
from pathlib import Path
import sys, json
ROOT = Path.cwd() if (Path.cwd() / 'cdlib').exists() else Path.cwd().parent
sys.path.insert(0, str(ROOT))
```

```python
import numpy as np
from PIL import Image
from cdlib.models.registry import load_model
from cdlib.inference import predict_pair
from cdlib.exports import evidence_zip
index = json.loads((ROOT / 'assets/demo/index.json').read_text())
pair = ROOT / 'assets/demo' / index[0]['dir']
a, b = [np.array(Image.open(pair / name).convert('RGB')) for name in ['A.png','B.png']]
model = load_model('tinycd', 'cpu', check_hash=True)
prob, seconds = predict_pair(model, a, b)
print('Output shape:', prob.shape, 'seconds:', seconds)
# Optional export: write evidence_zip(a, b, prob, 0.5, {'checkpoint': model.meta}) to a .zip file.
# The exhaustive interactive demo is: streamlit run app/app.py
# Actual Powai imagery follows USGS_WORKFLOW.md; never substitute this LEVIR pair for Powai.
```
