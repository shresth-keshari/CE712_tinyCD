```python
from pathlib import Path
import sys, json
ROOT = Path.cwd() if (Path.cwd() / 'cdlib').exists() else Path.cwd().parent
sys.path.insert(0, str(ROOT))
```

```python
import subprocess
import pandas as pd
RUN_EVALUATION = False  # Set True only to intentionally regenerate existing full-test artifacts.
if RUN_EVALUATION:
    for model in ['tinycd', 'bit', 'changeformer']:
        subprocess.run([sys.executable, 'scripts/evaluate.py', '--model', model, '--batch', '4'], cwd=ROOT, check=True)
rows = []
for model in ['tinycd', 'bit', 'changeformer']:
    result = json.loads((ROOT / f'results/eval_{model}_test.json').read_text())
    rows.append({'model': model, **result['metrics']})
pd.DataFrame(rows)
```
